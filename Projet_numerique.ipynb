L'objecitf du projet est de déterminer le potentiel électrique $V$ créé par une électrode $\Omega$ de dimensions finies, dans un échantillon, et nous allons y parvenir par analyse du problème.
Il s'agit du problème de Laplace, qu'on peut écire de la manière suivante :

$$\Delta V = 0, \text{ dans } \Omega$$
$$V \text{ constant sur l'électrode}$$
$$\frac{\partial V}{\partial n} = 0, \text{ dans } \partial\Omega, \text{ où } n \text{ est une variable de } V$$

Mais il n'existe pas de solution analytique à ce problème. 
Pourtant, il est possible de trouver une solution qui s'en approche. Pour cela, nous considérerons qu'en un point proche d'un bord de la plaque, les effets des autres bord seront négligeables. Nous prendrons donc une demi-plaque infinie, qui dans un repère à deux dimensions aura l'allure d'une demi-droite.

### 1. Première solution

#### 1.2. Théorie

Pour répondre à ce nouveau problème, nous partons de la fonction complexe $F$ définie de la manière suivant :
$$F(z) = z^\frac 12$$
Nous allons montrer que cette fonction permet de modéliser le potentiel autour de la plaque.
Comme $z$ et $F(z)$ sont des complexes, on peut écrire :
$$z=x+iy, \ \ \ \ F(x,y)=U(x,y)+iV(x,y)$$
où $U$ et $V$ sont deux fonctions réelles des variables $x$ et $y$.

Nous supposerons que le théorème mathématique suivant est vrai :
$$\frac {\partial U}{\partial x} = \frac {\partial V}{\partial y}$$
$$\frac {\partial U}{\partial y} = - \frac {\partial V}{\partial x}$$
On a alors :
$$\frac {\partial^2 U}{\partial x^2} = \frac {\partial}{\partial x} \left( \frac{\partial U}{\partial x} \right) = \frac {\partial}{\partial x} \left( \frac {\partial V}{\partial y} \right) = \frac {\partial^2 V}{\partial x \partial y}$$
$$\frac {\partial^2 U}{\partial y^2} = \frac {\partial}{\partial y} \left( \frac{\partial U}{\partial y} \right) = \frac {\partial}{\partial y} \left( -\frac {\partial V}{\partial x} \right) = -\frac {\partial^2 V}{\partial x \partial y}$$
$$\Longrightarrow \frac {\partial^2 U}{\partial x^2} + \frac {\partial^2 U}{\partial y^2} = \Delta U = 0$$
De même, 
$$\frac {\partial^2 V}{\partial x^2} = \frac {\partial}{\partial x} \left( \frac{\partial V}{\partial x} \right) = \frac {\partial}{\partial x} \left( -\frac {\partial U}{\partial y} \right) = -\frac {\partial^2 U}{\partial x \partial y}$$
$$\frac {\partial^2 V}{\partial y^2} = \frac {\partial}{\partial y} \left( \frac{\partial V}{\partial y} \right) = \frac {\partial}{\partial y} \left( \frac {\partial U}{\partial x} \right) = \frac {\partial^2 U}{\partial x \partial y}$$
$$\Longrightarrow \frac {\partial^2 V}{\partial x^2} + \frac {\partial^2 V}{\partial y^2} = \Delta V = 0$$

Les fonctions $U$ et $V$ remplissent donc la première condition au problème de Laplace. C'est d'ailleurs vrai pour toute fonction F complexe.
Pour les deux autres conditions, nous devrons exprimer explicitement $U$ et $V$.

On a :
$$F(z) = z^{\frac 12}$$
Comme $z= x + iy = \rho e^{i\theta}$ avec $\rho = (x^2 +y^2)^{\frac 12}$ et $\tan \theta = \frac yx,$
$$F(z) = (\rho e^{i\theta})^{\frac 12}$$
$$F(z) = \rho^{\frac 12} e^{i\frac \theta 2}$$
$$F(z) = \rho^{\frac 12} (\cos{\frac \theta 2} + i\sin{\frac \theta 2})$$
Or
$$\cos {\theta} = \cos(2 \cdot \frac{\theta}{2}) = 2\cos^2(\frac\theta 2) - 1 = \frac x\rho \Longrightarrow \cos{\frac \theta 2} = \left( \frac{x/\rho+1}{2} \right)^{\frac 12}$$
$$\cos {\theta} = 2\cos^2(\frac\theta 2) - 1 = 1 - 2\sin^2(\frac\theta 2)= \frac x\rho \Longrightarrow \sin{\frac \theta 2} = \left( \frac{1-x/\rho}{2} \right)^{\frac 12}$$
Donc
$$F(z) = \rho^{\frac 12} \left(\left( \frac{1 + x/\rho}{2} \right)^{\frac 12} + i\left( \frac{1-x/\rho}{2} \right)^{\frac 12}\right)$$
$$F(z) = \left( \frac{\rho+x}{2} \right)^{\frac 12} + i\left( \frac{\rho-x}{2} \right)^{\frac 12}$$
$$F(z) = \left( \frac{(x^2 +y^2)^{\frac 12}+x}{2} \right)^{\frac 12} + i\left( \frac{(x^2 +y^2)^{\frac 12}-x}{2} \right)^{\frac 12}$$
$$U(x,y) = \left( \frac{(x^2 +y^2)^{\frac 12}+x}{2} \right)^{\frac 12} \ ,\ \ \ V(x,y)=\left( \frac{(x^2 +y^2)^{\frac 12}-x}{2} \right)^{\frac 12}$$

Pour valider la deuxième consition, nous devons déterminer la position de la plaque dans l'espace. Nous avons dit précédemment qu'il s'agissait d'une demi-droite.
Dans le cas de $V$, c'est la demi-droite $y=0$ et $x≥0$. Dans le cas de $U$, c'est la demi-droite $y=0$ et $x≤0$.
Nous pouvons le vérifier de la manière suivante :
$$U(x,0) = \left( \frac{(x^2 +0^2)^{\frac 12}+x}{2} \right)^{\frac 12} = \left( \frac{|x|+x}{2} \right)^{\frac 12} = 0 \text{  si } x≤0$$
$$V(x,0) = \left( \frac{(x^2 +0^2)^{\frac 12}-x}{2} \right)^{\frac 12} = \left( \frac{|x|-x}{2} \right)^{\frac 12} = 0 \text{  si } x≥0$$

La troisième condition est difficile à montrer, car comme notre électrode est infinie, l'échantillon également, ce dernier ne possédant un bord qu'en $y=0$. On calcule les dérivées partielles de $U$ et $V$ :
$$\frac{\partial U}{\partial x} = \frac{\partial V}{\partial y} = \frac{\sqrt{2}}{4(x^2+y^2)^{\frac 12}}\cdot \frac{y}{(x^2+y^2)^{\frac 12}-x} = 0 \text{   si  } y= 0$$
$$\frac{\partial V}{\partial x} = -\frac{\partial U}{\partial y} = \frac{\sqrt{2}}{4(x^2+y^2)^{\frac 12}}\cdot \frac{y}{(x^2+y^2)^{\frac 12}+x} = 0 \text{   si  } y= 0$$

Donc les fonctions $U$ et $V$ vérifient bien les conditions de Laplace. Les deux fonctions peuvent donc représenter le potentiel électrique.

Par défaut, nous choisirons la fonction $V$ comme potentiel électrique. L'électrode se trouve alors en $x≥0$. La fonction $U$ peut alors représenter les lignes de champs électrique (nous pourrions le montrer mais ce n'est pas l'objectif ici).

#### 1.2. Graphes des Équipotentielles

Maintenant, nous allons coder dans python les courbes équipotentielles. Ci-dessous les fonctions $U$ et $V$, ainsi que les valeurs de $y$ pour $x$ et $V$ fixés.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def U(x,y):                                                  #Fonction U de x et y, qui représente les lignes de champs
    U = np.sqrt((np.sqrt(x ** 2 + y ** 2) + x) / 2)
    return U

def V(x, y):                                                 #Fonction V de x et y, qui représente le potentiel électrique
    V = np.sqrt((np.sqrt(x ** 2 + y ** 2) - x) / 2)
    return V
```

On peut commencer par tracer une équipotentielle pour un certain $V$ (vous pouvez modifier les valeurs de V et L, et observer les changements).

```python
V = 1
L = 10 # extension du domaine

def x(y,V):                                                  #Renvoie la position x en fonction de V et de y
    x = (y / 2 / V) ** 2 - V ** 2
    return x

y0 = 2 * V * np.sqrt(V ** 2 + L)
Y = np.arange(-y0, y0, 0.01)
X = x(Y, V)

plt.figure(figsize=(6, 6))
plt.plot([0,L],[0,0], color = 'black', label = 'plaque', linewidth = 3)
plt.plot(X, Y, color = 'blue', label = 'équipotentielle')
plt.title('Courbe équipotentielle V = '+ str(V))
plt.xlabel('position x')
plt.ylabel('position y')
plt.xlim(-L,L)
plt.ylim(-L,L)
plt.grid()
plt.legend()
plt.show()
```

Le programme suivant permet d'afficher plusieurs équipotentielles ainsi que les lignes de champs.

```python
L = 10
s = 0.01

def x(y,V):                                                  #Renvoie la position x en fonction de V et de y
    x = (y / 2 / V) ** 2 - V ** 2
    return x

plt.figure(figsize=(12, 5))                                  #taille du graphe

# N est le nombre de courbes tracées et m est l'écart de potentielle entre deux équipotentielles (n'hésitez pas à les modifier)
N = 6
m = 0.5
y0 = 2 * m * N * np.sqrt((m * N) ** 2 + L)
for i in range (1, N+1):
    V = m * i
    Y = np.arange(-y0, y0, s)
    X = x(Y, V)
    plt.plot(X, Y, 'b-')                                     #Équipotentielles
    plt.plot(-X, Y, 'r--')                                   #Lignes de champs

plt.plot([-L,0],[0,0], 'r--', label = 'Lignes de champs')    #Label lignes de champs 
plt.plot([0,L],[0,0], 'b-', label = 'Équipotentielles')      #Label équipotentielles
plt.plot([0,L],[0,0], color = 'black', label = 'Plaque', linewidth = 2)

plt.title(f'Équipotentielles et lignes de champs (les équipotentielles sont écartées de {m} Volt)')
plt.axis([-L,L,-10,1])
plt.xlabel('position x')
plt.ylabel('position y')
plt.grid()
plt.legend()
plt.show()
```

#### 1.3. Tolérance et Profils

Pour pouvoir comparer nos profils avec ceux des autres groupes (notamment le SP4), il faut entrer dans le deuxième programme les valeurs $H = 0.5$, et $L = 0.1$. Ces grandeurs ont été directement tirées du groupe SP4 et sont exprimées en mètres.

Nous allons ensuite tracer les profils de potentiel sous la plaque à différentes position $x$. Le problème, c'est que puisque notre plaque est infinie, elle n'a pas de longueur précise, et de fait, à coefficient près, nous n'avons que trois profils : le profil sous le bord en $x = 0$, le profil pour les $x$ positifs, et le profil pour les $x$ négatifs. Pour comparer nos profils avec les autres groupes, nous devons positionner le milieu $L$ de notre plaque à partir de la profondeur d'étude $H$ et d'une certaine tolérance $\tau$.

On sait qu'au milieu de la plaque, les courbes de potentiels devraient être horizontales. Donc les lignes de champs doivent être verticales. Pour une plaque infinie, cette situation n'arrive jamais, mais on peut s'en approcher. La tolérance donnera donc pour une certaine position $x$ l'écart entre les deux lignes de champs des profondeurs $H$ et $0$. Or dans notre analyse, étudier les lignes de champs aux positions $x$ revient à étudier le potentiel aux positions $-x$.

On obtient alors la relation suivante :
$$ \delta = V(-L, H) - V(-L, 0)$$
où $H$ est la profondeur d'étude, $L$ la demi-longueur de la plaque, et $\delta$ l'incertitude.
Cette relation est vraie dans le rectangle $[-L, L]\times[-H, H]$. Le calcule de la tolérance se fait de la manière suivante :
$$ \tau = \frac {\delta} {V(a, H)} $$
où $[-a, a]\times[-H, H]$ est le domaine d'étude, $a$ étant infiéreur ou égal à $L$.

On trace alors les profils de potentiel en $x=0$ (sous le bord de la plaque), en $x=L$ (sous le milieu de la plaque), en $x =\frac L2$ et en $x=-\frac L2$. Le programme calcule $\delta$ et $\tau$ à partir de $L$ et $H$. Il affiche également le profil en $x=-L$, pour vérifier la valeur de $\delta$.

```python
def V(x, y):
    V = np.sqrt((np.sqrt(x ** 2 + y ** 2) - x) / 2)
    return V

H = 0.05    # Profondeur d'étude
L = 0.1     # Demi-longueur de la plaque
a = 0.05    # Largeur d'étude (inférieur à L)

# Calcul de l'incertitude et de la tolérance
delta = V(-L, H) - V(-L, 0)
tau = delta / V(a, H)
print(delta, tau)

Y = np.linspace(0, H, 500)
V0 = V(0,Y)
V1 = V(L,Y)
V2 = V(a, Y)
V3 = V(-a, Y)
V4 = V(-L, Y)

# Courbe des profils
plt.plot(Y, V0, color = 'blue', label = 'x = 0')
plt.plot(Y, V1, color = 'red', label = 'x = L')
plt.plot(Y, V2, color = 'green', label = 'x = a')
plt.plot(Y, V3, color = 'black', label = 'x = -a')
plt.plot(Y, V4, color = 'yellow', label = 'x = -L')

# Mise en forme
plt.title(f'Profils de potentiel (L = {L:.2}, H = {H}, a = {a}, delta = {delta:.2}, tau = {tau:.2})')
plt.grid()
plt.legend()
plt.xlabel('position y')
plt.ylabel('Potentiel V')
plt.show()
```

### 2. Deuxième solution

#### 2.1. Théorie

Dans la première partie, nous avons considérer que l'électrode (la plaque) était sans épaisseur. Il peut être intéressant d'étudier un cas où l'épaisseur de l'électrode est prise en compte. Pour cela nous utiliserons la fonction $F(z) = z^{2/3}$, qui permet de modéliser le potentiel autour d'un coin infini.

Il sera plus simple de manipuler les puissances non entières en utilisant la forme polaire du nombre complexe z:

$$z = x + iy = \rho e^{iθ}$$

Où:
- $\rho = (x^2 + y^2)^{1/2}$
- $\tan \theta = \frac yx <=> θ = arctan(\frac yx)$ est l'argument

En utilisant les propriétés des puissances complexes, on applique l'exposant 2/3 à la forme polaire :

$$ F(z) = z^{2/3} = (\rho e^{iθ})^{2/3} $$
$$ F(z) = \rho^{2/3} e^{i \frac{2\theta}{3}} $$

D'après la formule d'Euler ($e^{i\alpha} = \cos\alpha + i\sin\alpha$), on peut séparer les parties réelle et imaginaire :

$$ F(z) = \rho^{2/3} \left( \cos\left(\frac{2\theta}{3}\right) + i\sin\left(\frac{2\theta}{3}\right) \right) $$

Par identification avec $F(z) = U(x, y) + iV(x, y)$, on trouve :
$$U(x, y) = \rho^{2/3} \cos\left(\frac{2\theta}{3}\right)$$
$$V(x, y) = \rho^{2/3} \sin\left(\frac{2\theta}{3}\right)$$

Nous n'allons pas simplifier d'avantage les formules de $U$ et $V$ (bien qu'il soit possible de réinjecter $\rho$ et $\theta$) car il est plus simple de coder avec les coordonnées polaires.

Nous savons que la première condition de Laplace est vérifié. Pour la deuxième :
$$ V(x, y) = 0 \Longleftrightarrow \frac{2\theta}{3} = n \pi $$
$$ V(x, y) = 0 \Longleftrightarrow \theta = \frac{3n \pi}2 $$
$$ V(x, y) = 0 \Longleftrightarrow \theta = 0 \text{ ou } \frac{3\pi}2 $$
Le coin se trouve donc entre les demi-droites $\{x=0, \, y ≤ 0 \}$ et $\{y=0, \, x ≥ 0 \}$.


#### 2.2. Équipotentielles

Maintenant, nous allons coder dans python les courbes équipotentielles. Ci-dessous les fonctions $U$ et $V$, ainsi que les valeurs de $y$ pour $x$ et $V$ fixés. L'approche cette fois ci est de construire un maillage, de calculer pour chaque point du maillage la fonction U et de tracer en reliant les points.

```python
import numpy as np
import matplotlib.pyplot as plt

# Valeur de l'équipotentielle
V = 2
L = 10

# Domaine angulaire
eps = 1e-4
theta = np.linspace(eps, 3*np.pi/2 - eps, 500)

# Rayon associé à V = 2
rho = (V / np.sin(2*theta/3))**(3/2)

# Coordonnées cartésiennes
x = rho * np.cos(theta)
y = rho * np.sin(theta)

# On élimine les valeurs trop grandes
mask = rho < 50
x = x[mask]
y = y[mask]

# Tracé
plt.figure(figsize=(6,6))

# Demi-droite noire (plaque)
plt.plot([0,L],[0,0], color='black', linewidth=3, label='coin')
plt.plot([0,0],[-L,0], color='black', linewidth=3)

# Equipotentielle
plt.plot(x, y, color='blue', label='équipotentielle')

plt.title(f'Courbe équipotentielle V = {V}')
plt.xlabel('position x')
plt.ylabel('position y')
plt.grid()
plt.legend()
plt.xlim(-L,L)
plt.ylim(-L,L)
plt.show()
```

#### 2.3. Effets de bord

**"Avec les mains"**


Dans cette partie nous allons reprendre ce que l'on fait dans la partie 1.3 et l'adapter à la recherche de ce qu'on pourrait appeler "*la zone de bord*". On a identifié précédemment 4 zones :

- $x≤0$
- $x = 0$ (le bord)
- $0≤x≤L$ 
- et x = L (le milieu)

Dans cette partie nous allons distingué la "zone de coin", c'est à dire une zone autour de $x=0$ dans laquelle l'effet du coin ne peut pas être négligé.

Autrement dit, l’idée principale de cette partie est de **quantifier à quelle distance du bord la plaque “se comporte comme infinie”**.
On cherche à partir de quelle position $x = \ell$ les effets de bord deviennent négligeables.

Toutes ses considérations sont bien sûr liées à la définition d'une tolérance $\tau$ pusique, tout comme le milieu n'est théoriquement jamais atteint, "on ne sort jamais de la *zone de coin*". La plaque est donc une sorte de coin infini...

<img src="fig2.png" height="500">

**Idée Physique**

Sous une plaque infinie, les lignes de champ sont parfaitement verticales et le potentiel varie uniquement avec y.

Près du bord, les lignes de champ “s’écartent” : c’est l’effet de bord. L’effet de bord devient négligeable lorsque la variation horizontale du potentiel devient plus petite que la tolérance $\tau$ (fixée).

$$\tau = V(-\ell,H) - V(-\ell,0)$$

Un mot sur la pertinence de ce critère :

* si $\tau$ est grand : les profils sont encore déformés
* si $\tau$ est petit : les équipotentielles sont presque horizontales => plus d’effet de bord

**Méthode numérique**

Petite explication de l’algorithme (similaire à ce qu'on a fait précédement) :

1. on fixe une profondeur d’étude H
2. on choisit une tolérance $\tau$
3. on recule progressivement dans la plaque ($x<0$)
4. on calcule : $V(x,H)-V(x,0)$
5. dès que cette différence devient inférieure à $\tau$, on considère que $\ell=-x$ est la distance minimale pour négliger les effets de bord

```python
def V(x, y):
    return np.sqrt((np.sqrt(x**2 + y**2) - x) / 2)

# Paramètres

H = 0.5

tau = 0.23

pas = 0.01

# Initialisation

x = 0
x_min = -100 # sécurité

while (abs(V(x, H) - V(x, 0)) > tau and x > x_min) :
    x -= pas

# Vérif

if x <= x_min:
    print("Tolérance non atteinte")
else:
    l = -x
    print("Effet de bord négligeable pour l =", L)
    print("Tolérance obtenue :", abs(V(x, H) - V(x, 0)))
```

La valeur trouvée de $\ell$ parait énorme surtout que le SP4 nous dit que L=0.1 ! On peut en conclure plusieurs choses:
- Premièrement que l'effet de bord ne se néglige pas à prtir d'une certaine profondeur (peu probable)
- Deuxièment que la valeur obtenue de L est incorecte : possible car on a trouvé une valeur de tau assez insatifaisante précédement
- Troisièmement que la méthode pour déterminer $\ell$ n'est pas bonne

Nous penchons pour une combinaison de la deuxième et la troisième : il se pourrait que le programme précédent converge naturellement vers L (cohérent avec la théorie).

On peut également raisonner dans l'autre sens. Ici on a fixé $\tau$ et calculé $\ell$ mais on peut aussi fixé $\ell$ et calculé $\tau$. Cela peut être très utile pour estimer "*l'erreur de bord*".

### 3. Comparaison des méthodes

On peut commencer par tracer les lignes d'équipotentielles pour une potentiel V en utilisant la première méthode, puis la deuxième méthode.

```python
# Valeur de l'équipotentielle
V = 0.5
L = 10

plt.figure(figsize=(10,8))

# Demi-droite noire (plaque)
plt.plot([0,L],[0,0], color='black', linewidth=3, label='électrode')
plt.plot([0,0],[0,-4], color='black', linestyle=":", linewidth=3, label='coin (selon méthode)')

def x(y,V):                                                  #Renvoie la position x en fonction de V et de y
    x = (y / 2 / V) ** 2 - V ** 2
    return x

y0 = 2 * V * np.sqrt(V ** 2 + L)
Y = np.linspace(0, y0, 500)
X = x(Y, V)

plt.plot(X, Y, color = 'orange', label = 'équipotentielle "plaque"')

# Domaine angulaire
eps = 1e-4
theta = np.linspace(eps, np.pi, 500)

# Rayon associé à V = 2
rho = (V / np.sin(2*theta/3))**(3/2)

# Coordonnées cartésiennes
x = rho * np.cos(theta)
y = rho * np.sin(theta)

# On élimine les valeurs trop grandes
mask = rho < 50
x = x[mask]
y = y[mask]

# Equipotentielle
plt.plot(x, y, color='green', label='équipotentielle "coin"')

plt.title(f'Comparaison des équipotentielles pour V = {V}')
plt.xlabel('position x')
plt.ylabel('position y')
plt.grid()
plt.legend()
plt.axis('equal')
plt.xlim(-L/2,L)
plt.show()
```

Le graphe précédent superpose les équipotentielles $V = 0.5$ obtenues par les deux modèles. Les deux courbes partent quasiment du même point : autour de (−0.25, 0) pour le modèle plaque et (−0.44,0) pour le modèle coin. Les valeurs sont si proches que les courbes sont indiscernables au voisinage du bord $x ≈ 0$.

En revanche, elles **divergent significativement à grand $x$**. Le comportement asymptotique de chaque courbe reflète directement l'exposant de la fonction génératrice. Pour le modèle plaque, à grand $x$ et petit $y/x$ :

$$V(x,y) = \sqrt{\frac{\sqrt{x^2+y^2}-x}{2}} \approx \sqrt{\frac{y^2}{4x}} = \frac{y}{2\sqrt{x}} = V_0 \implies y \sim 2V_0\sqrt{x}$$

Pour le modèle coin, à grand $x$ (petit $\theta$), avec $\sin\!\left(\frac{2\theta}{3}\right) \approx \frac{2\theta}{3}$ et $\rho \approx x$ :

$$V = \rho^{2/3}\sin\!\left(\frac{2\theta}{3}\right) \approx x^{2/3}\cdot\frac{2\theta}{3} = V_0 \implies \theta \approx \frac{3V_0}{2x^{2/3}} \implies y \approx x\theta \sim \frac{3V_0}{2}\,x^{1/3}$$

Donc pour la "*courbe plaque*" (courbe orange) qui suit une parabole $y \sim \sqrt{x}$ : l'équipotentielle s'éloigne de l'électrode de plus en plus vite au fur et à mesure qu'on s'en éloigne. La "*courbe coin*" (verte) qui suit une parabole $y \sim x^{1/3}$ reste quant à elle beaucoup plus proche de la surface de l'électrode par rapport à la *courbe plaque* (puisque 1/3 < 1/2).

Cette différence s'explique physiquement par la géométrie propre à chaque modèle. Dans le modèle coin, l'électrode possède une face inférieure 
(en $x=0, y≤0$) qui joue le rôle d'écran conducteur : elle "retient" les équipotentielles près de la surface et s'oppose à leur étalement. Le modèle plaque, lui, ne modélise qu'une électrode sans épaisseur, et le potentiel se propage donc librement dans tout le demi-espace supérieur.

**Conclusion sur les effets de coin dans la modélisation des équipotentielles** 

Dans cette  partie théorique concernant le problème de Laplace et la modélisation des équipotentielles autour de la plaque, les effets de coin jouent un rôle important, bien que localisé.
On a distingué trois zones :

1. Une zone très proche de l’électrode, où les effets de coin doivent être explicitement pris en compte.
2. Une zone intermédiaire, où ces effets deviennent négligeables : on peut alors considérer l’électrode comme infinie et sans épaisseur, en utilisant la première solution analytique.
3. Une zone éloignée, où la plaque est perçue comme finie, et où nos méthodes actuelles ne permettent plus de calculer le potentiel de manière efficace.
