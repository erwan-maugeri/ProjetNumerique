L'objecitf du projet est de déterminer le potentiel électrique $V$ autour d'une plaque $\Omega$ de dimensions finies, et nous allons y parvenir par analyse du problème.
Il s'agit du problème de Laplace, qu'on peut écire de la manière suivante :

$$\Delta V = 0, \text{ dans } \Omega$$
$$V \text{ constant sur la plaque}$$
$$\frac{\partial V}{\partial n} = 0, \text{ dans } \partial\Omega, \text{ où } n \text{ est une variable de } V$$

Mais il n'existe pas de solution analytique à ce problème. 
Pourtant, il est possible de trouver une solution qui s'en approche. Pour cela, nous considérerons qu'en un point proche d'un bord de la plaque, les effets des autres bord seront négligeables. Nous prendrons donc une demi-plaque infinie, qui dans un repère à deux dimensions aura l'allure d'une demi-droite.

### 1. Première solution

#### 1.2. Théorie

Pour répondre à ce nouveau problème, nous allons d'abord prendrons la fonction complexe $F$ définie de la manière suivant :
$$F(z) = z^\frac 12$$
Nous allons montrer que cette fonction permet de modéliser le potentiel autour de la plaque.
Comme $z$ et $F(z)$ sont des complexes, on peut écrire :
$$z=x+iy, \ \ \ \ F(x,y)=U(x,y)+iV(x,y)$$
où $U$ et $V$ sont deux fonctions réelles des variables $x$ et $y$.

Nous supposerons que le théorème mathématique suivant est vrai :
$$\frac {\partial U}{\partial x} = \frac {\partial V}{\partial y}$$
$$\frac {\partial U}{\partial y} = - \frac {\partial V}{\partial x}$$
On a alors :
$$\frac {\partial^2 U}{\partial x^2} = \frac {\partial}{\partial x} \left( \frac{\partial U}{\partial x} \right) = \frac {\partial}{\partial x} \left( \frac {\partial V}{\partial y} \right) = \frac {\partial^2 V}{\partial x \partial y}$$
$$\frac {\partial^2 U}{\partial y^2} = \frac {\partial}{\partial y} \left( \frac{\partial U}{\partial y} \right) = \frac {\partial}{\partial y} \left( -\frac {\partial V}{\partial x} \right) = -\frac {\partial^2 V}{\partial x \partial y}$$
$$\Longrightarrow \frac {\partial^2 U}{\partial x^2} + \frac {\partial^2 U}{\partial y^2} = \Delta U = 0$$
De même, 
$$\frac {\partial^2 V}{\partial x^2} = \frac {\partial}{\partial x} \left( \frac{\partial V}{\partial x} \right) = \frac {\partial}{\partial x} \left( -\frac {\partial U}{\partial y} \right) = -\frac {\partial^2 U}{\partial x \partial y}$$
$$\frac {\partial^2 V}{\partial y^2} = \frac {\partial}{\partial y} \left( \frac{\partial V}{\partial y} \right) = \frac {\partial}{\partial y} \left( \frac {\partial U}{\partial x} \right) = \frac {\partial^2 U}{\partial x \partial y}$$
$$\Longrightarrow \frac {\partial^2 V}{\partial x^2} + \frac {\partial^2 V}{\partial y^2} = \Delta V = 0$$

Les fonctions $U$ et $V$ remplissent donc la première condition au problème de Laplace. C'est d'ailleurs vrai pour toute fonction F complexe.
Pour les deux autres conditions, nous devrons exprimer explicitement $U$ et $V$.

On a :
$$F(z) = z^{\frac 12}$$
Comme $z= x + iy = \rho e^{i\theta}$ avec $\rho = (x^2 +y^2)^{\frac 12}$ et $\tan \theta = \frac yx,$
$$F(z) = (\rho e^{i\theta})^{\frac 12}$$
$$F(z) = \rho^{\frac 12} e^{i\frac \theta 2}$$
$$F(z) = \rho^{\frac 12} (\cos{\frac \theta 2} + i\sin{\frac \theta 2})$$
Or
$$\cos {\theta} = \cos(2 \cdot \frac{\theta}{2}) = 2\cos^2(\frac\theta 2) - 1 = \frac x\rho \Longrightarrow \cos{\frac \theta 2} = \left( \frac{x/\rho+1}{2} \right)^{\frac 12}$$
$$\cos {\theta} = 2\cos^2(\frac\theta 2) - 1 = 1 - 2\sin^2(\frac\theta 2)= \frac x\rho \Longrightarrow \sin{\frac \theta 2} = \left( \frac{1-x/\rho}{2} \right)^{\frac 12}$$
Donc
$$F(z) = \rho^{\frac 12} \left(\left( \frac{1 + x/\rho}{2} \right)^{\frac 12} + i\left( \frac{1-x/\rho}{2} \right)^{\frac 12}\right)$$
$$F(z) = \left( \frac{\rho+x}{2} \right)^{\frac 12} + i\left( \frac{\rho-x}{2} \right)^{\frac 12}$$
$$F(z) = \left( \frac{(x^2 +y^2)^{\frac 12}+x}{2} \right)^{\frac 12} + i\left( \frac{(x^2 +y^2)^{\frac 12}-x}{2} \right)^{\frac 12}$$
$$U(x,y) = \left( \frac{(x^2 +y^2)^{\frac 12}+x}{2} \right)^{\frac 12} \ ,\ \ \ V(x,y)=\left( \frac{(x^2 +y^2)^{\frac 12}-x}{2} \right)^{\frac 12}$$

Pour valider la deuxième consition, nous devons déterminer la position de la plaque dans l'espace. Nous avons dit précédemment qu'il s'agissait d'une demi-droite.
Dans le cas de $V$, c'est la demi-droite $y=0$ et $x≥0$. Dans le cas de $U$, c'est la demi-droite $y=0$ et $x≤0$.
Nous pouvons le vérifier de la manière suivante :
$$U(x,0) = \left( \frac{(x^2 +0^2)^{\frac 12}+x}{2} \right)^{\frac 12} = \left( \frac{|x|+x}{2} \right)^{\frac 12} = 0 \text{  si } x≤0$$
$$V(x,0) = \left( \frac{(x^2 +0^2)^{\frac 12}-x}{2} \right)^{\frac 12} = \left( \frac{|x|-x}{2} \right)^{\frac 12} = 0 \text{  si } x≥0$$

Pour la troisième condition, il faut calculer les dérivées partielles de $U$ et $V$ :
$$\frac{\partial U}{\partial x} = \frac{\partial V}{\partial y} = \frac{\sqrt{2}}{4(x^2+y^2)^{\frac 12}}\cdot \frac{y}{(x^2+y^2)^{\frac 12}-x} = 0 \text{   si  } y= 0$$
$$\frac{\partial V}{\partial x} = -\frac{\partial U}{\partial y} = \frac{\sqrt{2}}{4(x^2+y^2)^{\frac 12}}\cdot \frac{y}{(x^2+y^2)^{\frac 12}+x} = 0 \text{   si  } y= 0$$

Donc les fonctions $U$ et $V$ vérifient bien les conditions de Laplace. 
Nous choisirons la fonction $V$ comme potentiel électrique. 

#### 1.2. Équipotentielles

Maintenant, nous allons coder dans python les courbes équipotentielles. Ci-dessous les fonctions $U$ et $V$, ainsi que les valeurs de $y$ pour $x$ et $V$ fixés.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def U(x,y):                                                  #Fonction U de x et y, qui représente les lignes de champs
    U = np.sqrt((np.sqrt(x ** 2 + y ** 2) + x) / 2)
    return U

def V(x,y):                                                  #Renvoies le potentiel électrique V en fonction de x et y
    V = np.sqrt((np.sqrt(x ** 2 + y ** 2) - x) / 2)
    return V

def y(x,V):                                                  #Renvoie la position positive y en fonction de V et de x
    if V ** 2 + x <= 0:
        y = 0
    elif V ** 2 + x > 0:
        y = 2 * V * np.sqrt(V ** 2 + x)
    return y
```

On peut commencer par tracer une équipotentielle pour un certain V

```python
V = 2
L = 10
s = 0.01
```

```python
X = np.arange(-V**2, L, s)
Y0 = []
for i in X:
    Y0.append(y(i,V))
Y1 = []
for i in X:
    Y1.append(-y(i,V))

plt.plot([0,10],[0,0], color = 'black', label = 'plaque')
plt.plot(X, Y0, color = 'blue', label = 'équipotentiel')
plt.plot(X, Y1, color = 'blue')
plt.title('Courbe équipotentiel V = '+ str(V))
plt.legend()
plt.axis([-V**2,L,y(L,V),-y(L,V)])
plt.grid()
plt.show()
```

### 2. Deuxième solution

#### 2.1. Théorie

Une autre solution du problème est la fonction $F(z) = z^{2/3}$ . Cette fonction est adapté à la description du potentiel au bord de la plaque.

Il sera plus simple de manipuler les puissances non entières en utilisant la forme polaire du nombre complexe z:

$$z = x + iy = \rho e^{iθ}$$

Où:
- $\rho = (x^2 + y^2)^{1/2}$
- $θ = arctan(\frac yx)$ est l'argument

En utilisant les propriétés des puissances complexes, on applique l'exposant 2/3 à la forme polaire :

$$ F(z) = z^{2/3} = (\rho e^{iθ})^{2/3} $$
$$ F(z) = \rho^{2/3} e^{i \frac{2\theta}{3}} $$

D'après la formule d'Euler ($e^{i\alpha} = \cos\alpha + i\sin\alpha$), on peut séparer les parties réelle et imaginaire :
$$F(z) = \rho^{2/3} \left( \cos\left(\frac{2\theta}{3}\right) + i\sin\left(\frac{2\theta}{3}\right) \right)$$

Par identification avec $F(z) = U(x, y) + iV(x, y)$, on trouve :
$$U(x, y) = \rho^{2/3} \cos\left(\frac{2\theta}{3}\right)$$
$$V(x, y) = \rho^{2/3} \sin\left(\frac{2\theta}{3}\right)$$

En réinjectant les définitions de $\rho$ et $\theta$, on obtient les fonctions finales :
$$U(x, y) = (x^2 + y^2)^{1/3} \cos\left(\frac{2}{3} \arctan\left(\frac{y}{x}\right)\right)$$
$$V(x, y) = (x^2 + y^2)^{1/3} \sin\left(\frac{2}{3} \arctan\left(\frac{y}{x}\right)\right)$$

Nous aurons l'occasion de revenir sur l'efficacité de cette solution.

#### 2.2. Équipotentielles

Maintenant, nous allons coder dans python les courbes équipotentielles. Ci-dessous les fonctions $U$ et $V$, ainsi que les valeurs de $y$ pour $x$ et $V$ fixés. L'approche cette fois ci est de construire un maillage, de calculer pour chaque point du maillage la fonction U et de tracer en reliant les points.

```python
# Paramètres globaux
N = 8  # Nombre de lignes pour le maillage
L = 8 # Extension du domaine
V = 0.8 # Potentiel pour le tracé simple

def trace_z_23_simple(V):
    ''' tracé d'UNE SEULE ÉQUIPOTENTIELLE pour V donné '''
    fig, ax1 = plt.subplots()

    # On cherche l'ensemble des points tels que Im(z^(2/3)) = Constante
    # En polaire : r^(2/3) * sin(2*theta/3) = V0
    theta = np.linspace(0.01, 1.49 * np.pi, 500) # On évite les divisions par 0
    
    # r = (V0 / sin(2*theta/3))^(3/2)
    r = (V / np.sin(2 * theta / 3))**(1.5)
    
    x_eq = r * np.cos(theta)
    y_eq = r * np.sin(theta)
    
    ax1.plot(x_eq, y_eq, 'b-', linewidth=2, label=f"Equipotentielle V = {V}")
    ax1.plot([0, L], [0, 0], 'k-', linewidth=4, label="Plaque (modèle trait)")
    
    ax1.set_title(f"Tracé d'une seule équipotentielle ($z^{{2/3}}$)")
    ax1.set_xlim(-L, L)
    ax1.set_ylim(-L, L)
    ax1.grid(True, linestyle=':', alpha=0.6)
    ax1.legend()

    plt.show()

def trace_z_23_complet():
    ''' tracé de PLUSIEURS ÉQUIPOTENTIELLE et modélisation du COIN '''

    fig, ax2 = plt.subplots()
    # On utilise la transformation inverse : z = w^(3/2)
    # w = U + iV
    u_vals = np.linspace(0, L, N)
    v_vals = np.linspace(0, L, N)

    ax2.set_title(f"Maillage Conforme (N={N} points, $z = w^{{3/2}}$)")
    
    # Tracé des lignes à V constant (équipotentielles)
    for v in v_vals:
        u_fine = np.linspace(0, L, 100)
        w = u_fine + 1j * v
        z = w**(1.5)
        ax2.plot(z.real, z.imag, 'b-', linewidth=0.8)
        # On trace aussi le symétrique pour voir "autour" de la plaque
        ax2.plot(z.real, -z.imag, 'b-', linewidth=0.8)

    # Tracé des lignes à U constant (lignes de champ)
    for u in u_vals:
        v_fine = np.linspace(-L, L, 200)
        w = u + 1j * v_fine
        # On fait attention au signe pour la puissance complexe
        # z = w * sqrt(w)
        z = w**1.5
        ax2.plot(z.real, z.imag, 'r--', linewidth=0.8)

    # Dessin du coin (Erwan)
    # on reprend la fonction du haut et on trace seulement la première equipotentielle en noir
    v_fine = np.linspace(-L, L, 200)
    w = u_vals[0] + 1j * v_fine
    z = w**1.5
    ax2.plot(z.real, z.imag, 'black', linewidth=3, label="Coin")

    # réglage de l'affichage
    ax2.set_xlim(-L**1.5, L**1.5)
    ax2.set_ylim(-L**1.5, L**1.5)
    ax2.set_aspect('equal') # pour que l'angle paraisse droit (Erwan)
    ax2.grid(True)
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend()

    plt.show()

trace_z_23_simple(V)
trace_z_23_complet()
```

On peut aussi proposer une version plus complète de la fonction précédente qui prend pour argument l'angle de rotation du coin. La plaque est modélisée comme un rectangle parfait donc c'est bien un angle de 90 deg que l'on modélise

```python
def trace_z_23_advanced (angle=0):
    ''' Tracé de PLUSIEURS ÉQUIPOTENTIELLES et modélisation du COIN, avec rotation et symétrie
    
    Args:
        angle (float, optional): Angle de rotation en degrés. Par défaut, 0° (pas de rotation).'''

    # version adapté de la fonction précédente

    fig, ax2 = plt.subplots()
    theta = np.radians(angle)
    rotation = np.exp(1j * theta) # <<<<<<<<< ça se passe ici ...

    u_vals = np.linspace(0, L, N)
    v_vals = np.linspace(-L, L, N)  # On étend v_vals aux négatifs pour couvrir les deux côtés (symétrie)

    ax2.set_title(f"Maillage Conforme (N={N} points, rotation={angle}°)")

    # Tracé des lignes à V constant (équipotentielles) pour les deux côtés (repris de la fct précédente)
    for v in v_vals:
        u_fine = np.linspace(0, L, 100)
        w = u_fine + 1j * v
        z = w**1.5 * rotation
        ax2.plot(z.real, z.imag, 'b-', linewidth=0.8)

    # Tracé des lignes à U constant (lignes de champ) pour les deux côtés
    for u in u_vals:
        v_fine = np.linspace(-L, L, 200)
        w = u + 1j * v_fine
        z = w**1.5 * rotation
        ax2.plot(z.real, z.imag, 'r--', linewidth=0.8)

    # Dessin du coin (première équipotentielle en noir, Erwan)
    v_fine = np.linspace(-L, L, 200)
    w = u_vals[0] + 1j * v_fine
    z = w**1.5 * rotation
    ax2.plot(z.real, z.imag, 'black', linewidth=3, label="Coin")

    # Réglage de l'affichage (Erwan)
    max_val = L*2
    ax2.set_xlim(-max_val, max_val)
    ax2.set_ylim(-max_val, max_val)
    ax2.set_aspect('equal')
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend()

    plt.show()
```

```python
# Paramètres globaux
N = 15  # Nombre de lignes pour le maillage
L = 10 # Extension du domaine
V = 0.8 # Potentiel pour le tracé simple

trace_z_23_advanced (angle=0)
trace_z_23_advanced (angle=90)
trace_z_23_advanced (angle=135)

help(trace_z_23_advanced)
```

### 3. Exemple d'une solution qui ne marche pas

À observer les 2 cas précédent, on pourrait penser qu'il est possible de proposer n'importe quelle fonction complexe F(z). 

Nous allons maintenant considérer $F(z) = z^{3/2}$ et montrer que cette fonction ne permet pas de modéliser le potentiel autour de la plaque. Pour ce faire nous allons (comme précédement):
- Détermnier les fonctions U et V
- Vérifier les conditions de Laplace

Pour exprimer $U$ et $V$, on utilise le fait que :
$$z^{3/2} = z \cdot z^{1/2}$$

En posant $U_{1/2}$ et $V_{1/2}$ les fonctions obtenues pour la première solution ($z^{1/2}$), on a :
$$F(z) = (x + iy) \cdot (U_{1/2} + iV_{1/2})$$
$$F(z) = x U_{1/2} + i x V_{1/2} + i y U_{1/2} - y V_{1/2}$$
$$F(z) = (x U_{1/2} - y V_{1/2}) + i(x V_{1/2} + y U_{1/2})$$

Par identification, on obtient l'expression de $U(x,y)$ :
$$U(x,y) = x \left( \frac{(x^2 +y^2)^{1/2}+x}{2} \right)^{1/2} - y \left( \frac{(x^2 +y^2)^{1/2}-x}{2} \right)^{1/2}$$

et l'expression de $V(x,y)$ (Potentiel électrique) :
$$V(x,y) = x \left( \frac{(x^2 +y^2)^{1/2}-x}{2} \right)^{1/2} + y \left( \frac{(x^2 +y^2)^{1/2}+x}{2} \right)^{1/2}$$

Or $\rho = (x^2 + y^2)^{1/2}$ donc on peut simplifier l'écriture:

$$U(x,y) = x \left( \frac{\rho + x}{2} \right)^{1/2} - y \left( \frac{\rho - x}{2} \right)^{1/2}$$

$$V(x,y) = x \left( \frac{\rho - x}{2} \right)^{1/2} + y \left( \frac{\rho + x}{2} \right)^{1/2}$$

$$ \text{(noter que l'écriture $\rho$ est une bonne pratique pour la programmation)} $$

Il nous reste à verifier qu'elle vérifie les conditions du problème de Laplace:

Les 2 premières conditions fonctionnent, d'ailleurs elles fonctionnent pour toute fonction complexe U et V.

Pour la troisième condition (nulité des dérivées partielles), on calcule $\frac{\partial U}{\partial x}$ :

$$\frac{\partial U}{\partial x} = \frac{\sqrt{2}}{4(x^2+y^2)^{\frac 12}}\cdot \frac{y}{(x^2+y^2)^{\frac 12}-x} x + \left( \frac{(x^2 +y^2)^{1/2}+x}{2} \right)^{1/2} + ...$$

condition "$= 0$ si $y=0$" (nulité sur la plaque) non vérifiée car si y = 0:

$$\frac{\partial U}{\partial x} \quad = \quad 0 \cdot x + \left( \frac{(x^2 +0)^{1/2}+x}{2} \right)^{1/2} + \quad... \quad≠\quad 0 $$

(on a besoin d'une condition supplémentaire sur x)

**Conclusion:** Cette solution modélise ***UN*** problème de Laplace mais on a montré qu'il ne s'agisait pas ***DU*** problème de Laplace auquel on s'intéresse. On ne peux pas prendre n'importe quelle fonction F(z) !
